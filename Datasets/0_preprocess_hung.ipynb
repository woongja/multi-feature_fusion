```python
import pandas as pd

df = pd.read_csv("/home/woongjae/noise-tracing/multi-feature_fusion/Datasets/meta.csv")

# 라벨 변환
df['label1'] = df['label1'].replace({
    'high_pass_filter': 'band_pass_filter',
    'low_pass_filter': 'band_pass_filter',
    'pitch_shift': 'manipulation',
    'time_stretch': 'manipulation'
})

# 고유 노이즈 타입 목록 재생성 (중복제거)
noise_types = df['label1'].unique()

protocol_entries = []
n = 400
for noise_type in noise_types:
    bona = df[(df['label1'] == noise_type) & (df['label2'] == 'bonafide')]
    spoof = df[(df['label1'] == noise_type) & (df['label2'] == 'spoof')]

    bona_sample = bona.sample(n=min(n, len(bona)), random_state=42)
    spoof_sample = spoof.sample(n=min(n, len(spoof)), random_state=42)

    protocol_entries.append(bona_sample[['file_path', 'label2', 'label1']])
    protocol_entries.append(spoof_sample[['file_path', 'label2', 'label1']])

protocol_df = pd.concat(protocol_entries, ignore_index=True)

protocol_df.to_csv("protocol_noise_custom.csv", index=False, header=False)
```
